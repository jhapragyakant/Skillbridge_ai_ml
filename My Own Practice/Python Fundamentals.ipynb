Data Types

```python
# Primitive / basic data types in Python

integer_value = 42
float_value = 3.14
complex_value = 2 + 5j
string_value = "Hello, Python"
boolean_value = True
none_value = None
bytes_value = b"Hello"

print(integer_value, type(integer_value))
print(float_value, type(float_value))
print(complex_value, type(complex_value))
print(string_value, type(string_value))
print(boolean_value, type(boolean_value))
print(none_value, type(none_value))
print(bytes_value, type(bytes_value))

# Infinity is represented as positive or negative infinity of type float
positive_infinity = float("inf")
negative_infinity = float("-inf")

print(positive_infinity)
print(negative_infinity)

# Arithmetic examples
print(positive_infinity + 100)  # inf
print(positive_infinity * 0)    # nan, because infinity times zero is undefined
# print(10.0 / 0.0)               # This raises ZeroDivisionError
print(1e308 * 1e308)            # inf
```

```python
print(1e309) # This is considered to be infinity by python
print(1e308)
```

```python
# More complex data types

# List: ordered and mutable
example_list = [1, 2, "Python", True]
print("List:", example_list)
print("List type:", type(example_list))

# Tuple: ordered and immutable
example_tuple = (1, 2, "Python", True)
print("Tuple:", example_tuple)
print("Tuple type:", type(example_tuple))

# Dictionary: key-value pairs
example_dict = {
    "name": "Pragya",
    "age": 25,
    "skills": ["Python", "Machine Learning"]
}
print("Dictionary:", example_dict)
print("Dictionary type:", type(example_dict))

# Set: unordered collection of unique values
example_set = {1, 2, 3, 3, 4}
print("Set:", example_set)
print("Set type:", type(example_set))
```

```python
# input() always returns data as a string
name = input("Enter your name: ")
print("Name:", name)
print("Name type:", type(name))

# Typecast string input to other data types
age = int(input("Enter your age: "))
height = float(input("Enter your height in meters: "))
is_student = input("Are you a student? (yes/no): ").strip().lower() == "yes"

print(f"\nHello {name}!")
print(f"You are {age} years old.")
print(f"Your height is {height} meters.")
print(f"Student status: {is_student}")

print("Age type:", type(age))
print("Height type:", type(height))
print("Student status type:", type(is_student))

# Use the converted values
age_after_one_year = age + 1
print(f"Next year, you will be {age_after_one_year} years old.")
```

Operators

```python
# 1. Arithmetic Operators
a = 15
b = 4

print("Arithmetic operators:")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Modulus:", a % b)
print("Exponentiation:", a ** b)

# 2. Relationship (Comparison) Operators
print("\nRelationship operators:")
print("a == b:", a == b)
print("a != b:", a != b)
print("a > b:", a > b)
print("a < b:", a < b)
print("a >= b:", a >= b)
print("a <= b:", a <= b)

# 3. Logical Operators
is_adult = age >= 18
is_learning = True

print("\nLogical operators:")
print("is_adult and is_learning:", is_adult and is_learning)
print("is_adult or is_learning:", is_adult or is_learning)
print("not is_learning:", not is_learning)

# 4. Bitwise Operators
x = 10
y = 6

print("\nBitwise operators:")
print("x & y:", x & y)
print("x | y:", x | y)
print("x ^ y:", x ^ y)
print("~x:", ~x)
print("x << 1:", x << 1)
print("x >> 1:", x >> 1)

# 5. Membership Operators
skills = ["Python", "Machine Learning", "SQL"]

print("\nMembership operators:")
print("'Python' in skills:", "Python" in skills)
print("'Java' not in skills:", "Java" not in skills)
print("'name' in example_dict:", "name" in example_dict)

# 6. Assignment Operators
score = 10
print("\nAssignment operators:")
print("Initial score:", score)

score += 5
print("After += 5:", score)

score -= 3
print("After -= 3:", score)

score *= 2
print("After *= 2:", score)

score /= 4
print("After /= 4:", score)

score %= 3
print("After %= 3:", score)

score = 10
score **= 2
print("After **= 2:", score)

score //= 3
print("After //= 3:", score)
```
