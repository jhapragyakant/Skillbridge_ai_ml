# Write a program to take 3 digit number from user and sum its digit

```python
num = int(input("Enter a three digit number "))

a = num % 10                        #Gives unit place digit

num = num // 10                     #Gives number without unit place digit

b = num % 10                        #Gives ten's place digit

num = num //10                      #Gives number without ten's place digit

print("The digits are ",num, b, a)
print("The sum of digits is ",a+b+num)
```
